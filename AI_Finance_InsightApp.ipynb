```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    score = analyzer.polarity_scores(text)

    if score['compound'] >= 0.05:
        sentiment = "Positive"
    elif score['compound'] <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return sentiment, score
```

```python
article = """
2
Apple reported stronger-than-expected quarterly earnings.
3
Revenue increased by 15% and investors responded positively.
4
The company expects continued growth in AI-powered products.
5
"""
```

```python
sentiment, score = analyze_sentiment(article)

print("Sentiment:", sentiment)
print("Score:", score)
```

```python
import pandas as pd

# Create a dictionary representing our data row
data = {
    "article": [article.strip()],
    "sentiment": [sentiment],
    "compound_score": [score['compound']],
    "positive_score": [score['pos']],
    "neutral_score": [score['neu']],
    "negative_score": [score['neg']]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Save to a CSV file
csv_path = "sentiment_results.csv"
df.to_csv(csv_path, index=False)

print(f"Results successfully saved to {csv_path}!")
print(df)
```
