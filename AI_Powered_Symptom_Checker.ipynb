```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
# Download NLP resources
print("Downloads Complete")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
# Create medical symptom database
print("Libraries Loaded Successfully")
```

```python
# Symptom database
# Format:
# "symptoms": ("condition", "triage level")
# NLP preprocessing function
symptom_database = {
    "headache fever sore throat": ("Flu", "Urgent Care"),
    "cough sore throat": ("Common Cold", "Self Care"),
    "chest pain shortness breath": ("Possible Heart Condition", "Emergency"),
    "abdominal pain vomiting": ("Possible Gastroenteritis", "Urgent Care"),
    "runny nose sneezing": ("Allergies", "Self Care"),
    "high fever body aches fatigue": ("Influenza", "Urgent Care"),
    "difficulty breathing chest tightness": ("Possible Respiratory Emergency", "Emergency")
}

print("Symptom database loaded successfully.")
```

```python
def process_symptoms(text):
    # Convert text into tokens
    tokens = word_tokenize(text.lower())

    # Remove common stop words
    stop_words = set(stopwords.words('english'))

    tokens = [
        word for word in tokens
        if word.isalpha()
        and word not in stop_words
    ]

    # Convert words to base form
    lemmatizer = WordNetLemmatizer()

    lemmatized_tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return lemmatized_tokens
```

```python
sample = "I have a severe headache and high fever"
result = process_symptoms(sample)
print("Processed tokens:", result) # User input
```

```python
def analyze_symptoms(symptoms):# Import required libraries
    processed = process_symptoms(symptoms)
    best_match = None
    highest_score = 0

    for phrase, (condition, triage) in symptom_database.items():
        keywords = phrase.split()
        score = sum(
            1 for word in processed
            if word in keywords
        ) # Display results

        if score > highest_score:
            highest_score = score
            best_match = (condition, triage)

    if best_match:
        return best_match

    return (
        "Unknown Condition",
        "Consult a Healthcare Provider"
    )
```

```python
user_symptoms = "I have a headache, fever, and sore throat"
condition, triage = analyze_symptoms(user_symptoms)
# Import required libraries
print("Possible Condition:", condition)
print("Triage Level:", triage)
```

```python
user_input = input("Enter your symptoms: ")
condition, triage = analyze_symptoms(user_input)

print("\nResults")
print("----------------------")
print("Possible Condition:", condition)
print("Triage Level:", triage)

print("\nDisclaimer:")
print("This tool is for educational purposes only and is not a substitute for professional medical advice.")
```
